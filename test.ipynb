```python
def product_rule_derivative(f_coeffs: list, g_coeffs: list) -> list:

    while(len(f_coeffs) < 3):
        f_coeffs.append(0)
    while(len(g_coeffs) < 3):
        f_coeffs.append(0)

    
    result = []
    result.append(f_coeffs[0] * g_coeffs[4] + f_coeffs[1] * g_coeffs[3])
    result.append(f_coeffs[0] * g_coeffs[5] + f_coeffs[1] * g_coeffs[4] + f_coeffs[2] * g_coeffs[3])
    result.append(f_coeffs[1] * g_coeffs[5] + f_coeffs[2] * g_coeffs[4])
    result.append(f_coeffs[2] * g_coeffs[5])
    counter = len(result)
    for i in reversed(result):
        if(result == 0):
            counter = counter - 1
    res = result[0: counter]
    if(len(res) == 0):
        return 0
    return res

product_rule_derivative([1,2],[3,4])
```
